```python
# Import necessary libraries
import pytesseract
from PIL import Image
from google.colab import files
```

```python
# get image of the dry food we want to check
image_path='/content/Screenshot 2026-03-15 at 12.38.51 AM.png'
```

```python
# Open the image and perform OCR
extracted_text = pytesseract.image_to_string(Image.open(image_path))

# Print the extracted text to check the image recognition
print("Extracted Text:")
print(extracted_text)
```

```python
# list of allergies
sentence="""Artificial beef
flavour
Beef
Beef bone
Brussels sprout
Chicken fat
Chicken meal
Corned beef
Crab
Ground lamb
Ground turkey
Oranges
Parsnips
Sweet potato
Vinegar-malt
Agar
Apples
Artificial bacon
flavour
Brie
Cheddar cheese
Coconut oil
Ground beef
Haddock
Palm oil
Pasta
Pig ear
Popcorn
Sodium acid
phosphate"""
allergies=sentence.split('\n')
print(allergies)
```

```python
#checking each allergy to the ingredients of the food, if something matches,print the ingredient
matching_items = [allergy_item for allergy_item in allergies if allergy_item.lower() in extracted_text.lower()]
print(f"Matching items: {matching_items}")
```
